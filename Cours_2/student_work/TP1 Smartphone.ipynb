### Transformation en Valeur exploitable 🔃
Bon la quasi-totalité est en format objet (Des string avec unité) ce qui n'est pas très pratique pour les stats. Transformons ça en Int et consorts.

1) Cas un peu spécial à cause des NaN, j'utilise Strip

```python
df_smartphone['Mobile Weight'] = df_smartphone['Mobile Weight'].str.rstrip('g').astype(float)
df_smartphone.rename(columns={'Mobile Weight': 'Mobile Weight (g)'}, inplace=True)
df_smartphone.head(10)  # Afficher les premières lignes pour vérifier
```

```python
def remove_units(column, unit):
    column= column.str.replace(',', '', regex=False)
    column= column.str.replace('¥', 'CNY', regex=False)
    return column.str.replace(unit, '', regex=False).str.strip().replace('Not available', pd.NA).apply(lambda x: float(x) if pd.notna(x) else x)

df_smartphone['Launched Price (China)'] = remove_units(df_smartphone['Launched Price (China)'], 'CNY')
df_smartphone['Launched Price (Dubai)'] = remove_units(df_smartphone['Launched Price (Dubai)'], 'AED')
df_smartphone['Launched Price (USA)'] = remove_units(df_smartphone['Launched Price (USA)'], 'USD')
df_smartphone['Launched Price (India)'] = remove_units(df_smartphone['Launched Price (India)'], 'INR')
df_smartphone['Launched Price (Pakistan)'] = remove_units(df_smartphone['Launched Price (Pakistan)'], 'PKR')
df_smartphone['Battery Capacity'] = remove_units(df_smartphone['Battery Capacity'], 'mAh')
# df_smartphone['RAM'] = remove_units(df_smartphone['RAM'], 'GB') 
df_smartphone.tail(10)  # Afficher les premières lignes pour vérifier
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Préparation des données
weight_data = df_smartphone[['Mobile Weight (g)']].dropna()  # Supprimer les NaN
scaler = StandardScaler()
weight_scaled = scaler.fit_transform(weight_data)
```

```python
inertia = []
K = range(1, 11)  # Tester de 1 à 10 clusters

for k in K:
    kmeans_test = KMeans(n_clusters=k, random_state=42)
    kmeans_test.fit(weight_scaled)
    inertia.append(kmeans_test.inertia_)

# Tracer le graphe de la méthode du coude
plt.figure(figsize=(8, 6))
plt.plot(K, inertia, 'bo-', markersize=8)
plt.title("Méthode du coude pour déterminer le nombre optimal de clusters")
plt.xlabel("Nombre de clusters")
plt.ylabel("Inertie intra-cluster")
plt.grid()
plt.show()
```

```python
# Appliquer KMeans
kmeans = KMeans(n_clusters=2, random_state=42)
weight_data['Cluster'] = kmeans.fit_predict(weight_scaled)



# Visualisation des clusters
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=weight_data,
    x='Mobile Weight (g)',
    y=[0] * len(weight_data),  # Ajout d'une dimension fictive pour visualiser
    hue='Cluster',
    palette='viridis',
    alpha=0.7
)
plt.title('Clusters basés sur le poids des smartphones')
plt.xlabel('Poids (g)')
plt.show()
```

### Cherchons des corrélation diverse : 

```python
# Filtrer les smartphones pesant moins de 300 g
filtered_df = df_smartphone[df_smartphone['Mobile Weight (g)'] < 300]

# Calculer la matrice de corrélation pour les données filtrées
correlation_matrix_filtered = filtered_df.corr()

# Tracer une heatmap pour les données filtrées
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix_filtered, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Matrice de corrélation des valeurs numériques (Poids < 300g)")
plt.show()
```

Ce qu'on savais déja ce voit bien. On peut remarquer cependant 2 choses :

- Il y a aussi corrélation pas si élevé entre les conversion de prix, montrant des prix régionaux modifié manuellement.
- Le poids est légérement corrélé a tout.

#### Avec les caméra arrière (en partie et sans cas spéciaux)

```python
# Filtrer les appareils de moins de 320 g
filtered_df = df_smartphone[df_smartphone['Mobile Weight (g)'] < 320]

# Transformation de la colonne "Back Camera" en une valeur numérique
def extract_megapixels(camera_column):
    return camera_column.str.extract(r'(\d+\.?\d*)').astype(float)

filtered_df['Back Camera (MP)'] = extract_megapixels(filtered_df['Back Camera'])

# Calcul de la corrélation entre le prix de lancement (USA) et la caméra arrière
correlation_back_camera = filtered_df['Back Camera (MP)'].corr(filtered_df['Launched Price (USA)'])
print(f"Corrélation entre la caméra arrière et le prix de lancement (USA) : {correlation_back_camera:.2f}")

# Graphique
plt.figure(figsize=(10, 6))
sns.regplot(
    data=filtered_df,
    x='Back Camera (MP)',
    y='Launched Price (USA)',
    scatter_kws={'alpha': 0.7},
    line_kws={'color': 'red', 'linewidth': 2}
)
plt.title('Relation entre la caméra arrière (MP) et le prix de lancement (USA) (Poids < 320g)')
plt.xlabel('Caméra arrière (MP)')
plt.ylabel('Prix de lancement (USA)')
plt.tight_layout()
plt.show()
```

Décidement les corrélation sont faibles !
